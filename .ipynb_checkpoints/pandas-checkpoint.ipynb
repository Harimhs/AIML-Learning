# Adding a new column

```python
print("if need a same static value for all rows: ")
df['language score']= 70
print(df.head())

print("if need a column from an array values: ")
arr= np.arange(len(df)) #stores 0-999 elements (length of df)
df['language score']= arr
print(df.head())

print("if need a column from an array values in random: ")
arr= np.random.randint(1, 100, len(df))
df['language score']= arr
print(df.head())
```

# Adding multiple columns

```python
arr= np.random.randint(1, 100, len(df))
arr1= np.random.randint(1, 100, len(df))

#series represents a column
score1= pd.Series(arr, index=np.arange(0, len(df)))
score2= pd.Series(arr1, index=np.arange(0, len(df)))

#using assign() function
df= df.assign(score_1= score1, score_2= score2)
df.head()
```

# Inseting column

```python
df= pd.read_csv('demo.csv')

arr= np.random.randint(1, 100, len(df))
series= pd.Series(arr, index=np.arange(0, len(df)))
df.insert(1, 'value', series)
print(df.head())
```

# Math Operations

```python
print(df['math score'].sum())
print(df['value'].max())
print(df['value'].std())
print(df.describe()) #all operations
```

# Operations in rows

```python
#adding rows
df['math score']+df['writing score']+df['reading score']

#adding rows and assign the result to a column
df['average']= (df['math score']+df['writing score']+df['reading score'])/3
print(df.head())
```

# Values count

```python
#count a column
print(df['gender'].count())
#counting a column elements by category
print(df['gender'].value_counts())
#return the relative frequency
print(df['gender'].value_counts(normalize= True))
```

# Sorting

```python
# sorting by a column
df.sort_values('math score')
#descending order
df.sort_values('math score', ascending= False)
#sorting by multiple columns
df.sort_values(['math score', 'reading score'], ascending= False)
#sorting in inplace
df.sort_values(['math score', 'reading score'], ascending= False, inplace= True)
```

# Query, loc[], iloc[]

```python
data= {
    'Name': ['Gun', 'Zack', 'Daniel', 'Vasco'],
    'Age': [28, 24, 23, 24],
    'Department': ['IT', 'HR', 'Sales', 'Design'],
    'Salary': [50000, 40000, 40000, 450000]
}
df= pd.DataFrame(data)

print('Task 1: select rows where Age>25.')

#using query()
print(df.query('Age>25'))

#using loc[]
print(df.loc[df['Age']>25])

#using iloc[]
print('Cannot use iloc for condition-based filtering. iloc is strictly position-based, not value-based.')

print('Task 2: Get Salary of employee at index 3')

#using query()
print(df.query('index==3'))

#using loc[]
print(df.loc[3, 'Salary'])

#using iloc[]
print(df.iloc[3, 3])

print('Task 3: Select Name and Department of first 3 employees')

#using loc[]
print(df.loc[0:2, ['Name', 'Department']])

#using iloc[]
print(df.iloc[0:3, 0:2])
```
