```python
import pandas as pd
import numpy as np
```

# Series

```python
list= np.arange(5)
data= pd.Series(list)
print(data)

list1 = np.arange(1,5).reshape((2, 2))
# data1= pd.Series(list1) Series won't work on matrix
data1= pd.Series(list1.flatten())
print(data1)
```

# Dataframe

```python
#creating dataframe using array
arr= np.arange(1,5).reshape(2,2)
df= pd.DataFrame(arr, index=['row1', 'row2'], columns=['col1', 'col2'])
print(df)

print()

#using list
list= [["ORV",565], ["LOTM", 1540]]
df1= pd.DataFrame(list, index=[1, 2], columns=["Novels", "Chapters"])
print(df1)

print()

#using dictionary
novels= ["ORV", "LOTM", "TBATE", "TGED"]
chapters= [565, 1540, 730, 230]
dict= {"Novels": novels, "Chapters":chapters}
df2= pd.DataFrame(dict, index=[1,2,3,4])
print(df2)

print()

#using csv
df3= pd.read_csv('demo.csv')
#print first 5 rows
print(df3.head()) 
#print last 5 rows
print(df3.tail())
#print first n rows 
print(df3.head(10))
#print last n rows 
print(df3.tail(10))
#print shape of csv rows, cols
print(df3.shape)
#print n rows
pd.set_option('display.max_rows', 1000)
# df3
```

# Selecting single column - 2 syntax

```python
#selecting by []
print(df['gender'])
print(df['gender'].head())
print(type(df['gender']))

#selecting by .
print(df.gender)
print(df.gender.tail())
#error df.math score 
```

# Adding a new column

```python
print("if need a same static value for all rows: ")
df['language score']= 70
print(df.head())

print("if need a column from an array values: ")
arr= np.arange(len(df)) #stores 0-999 elements (length of df)
df['language score']= arr
print(df.head())

print("if need a column from an array values in random: ")
arr= np.random.randint(1, 100, len(df))
df['language score']= arr
print(df.head())
```

# Adding multiple columns

```python
arr= np.random.randint(1, 100, len(df))
arr1= np.random.randint(1, 100, len(df))

#series represents a column
score1= pd.Series(arr, index=np.arange(0, len(df)))
score2= pd.Series(arr1, index=np.arange(0, len(df)))

#using assign() function
df= df.assign(score_1= score1, score_2= score2)
df.head()
```
