# List

```python
list= [10, 20, 30, 40]
#length
print(len(list))
#slicing
print(list[:4])
#modify
list[1]= 15
print(list[1])

#adding elements (append, insert)
list.append(50)
print(list)
pos= 1
element= 20
list.insert(pos, element)
print(list)
list1= [50, 60, 70, 80]
nested_list= [list+list1]
print(nested_list)

#removing elements (remove, pop, del)
list.remove(15) #required direct element
print(list)
list.pop(-1) #index position
print(list)
del list[0]
print(list)

#sorting the list
list.sort()
print(list)
list.sort(reverse= True)
print(list)
```

# Dictionary

```python
dict= {"name":"hari", "age":22}
print(dict.keys())
print(dict.values())
print(dict.items())

#adding element
dict["height"]= 1.8
print(dict)

#update value
dict.update({"height":2.1})
print(dict)

#removing, only keys can be perfomed for removal not index or values
dict.pop("height")
print(dict)
del dict["age"]
print(dict)
dict.clear()
print(dict)
```

# Set

```python
a= {1, 2, 3}
b= {3, 4, 5}
for item in a:
  print(item)
a.add(0)
a.update([11, 12]) #multiple elements
a.remove(11)
a.discard(12)
a.pop() #remove random element
for item in a:
  print(item)
print(a|b) #union
print(a&b) #intersect
print(a-b) #difference
print(a^b) #symmetric
```

# For Loop

```python
list= ["AI", "ML", "DS", "DSA"]
for item in list:
  print(item)

#enumerate + for loop
for i, item in enumerate(list):
  print(i)
  print(item)

#dictionary + for loop
dict= {"name":"hari", "age": 22}
for key, value in dict.items():
  print(key)
  print(value)
```
