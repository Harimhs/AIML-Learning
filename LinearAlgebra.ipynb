```python
import numpy as np
```

# **Vector Length aka Magnitude or Norm**

```python
def vector_length(vector):
    root_sqr_sum=0
    for i in vector:
        root_sqr_sum+= i*i
    return np.sqrt(root_sqr_sum)
    
# v= np.array([2,5,3])
# print(vector_length(v))
```

# **Vector Direction**

```python
def vector_dir(vector):
    length= vector_length(vector)
    dir=[float(i/length) for i in vector]
    return dir

v= np.array([2,5,3])
print(vector_dir(v))
```

# **Vector Addition**

```python
def check_size(*args):
    for i in range(len(args)-1):
        if len(i!=len(i+1)):
            return False
    return True

def vector_add(*args):
    list_vector= []
    for i in args:
        list_vector.append(i)
    if(check_size(list_vector)):
        size= len(list_vector[0])
        add=[]
        for i in range(size):
            sum=0
            for j in list_vector:
                sum+= j[i]
            add.append(sum)
    else:
        print("Not equal vector dimensions!")
    return add

print(vector_add([1,2], [3,4], [5,6]))
    
```

# **Dot Product**

```python
def dot_prod(v1, v2):
    if len(v1)!=len(v2):
        return False
    sum=0
    for a, b in zip(v1, v2):
        prod = a*b
        sum+=prod
    return sum

print(dot_prod([1,2], [3,4]))
        
```

# **Scalar Multiplication**

```python
def scalar_mul(s, v):
    for i in range(len(v)):
        v[i]= s*v[i]
    return v

print(scalar_mul(2, [3, 5]))
```

# **Distance Between Vectors**

```python
def distance(v1, v2):
    if len(v1)!=len(v2):
        return False
    sum=0
    for i, j in zip(v1, v2):
        temp= i-j
        sum+=temp*temp
    sum= np.sqrt(sum)
    return sum

print(distance([1,2], [3,4]))
    
```
